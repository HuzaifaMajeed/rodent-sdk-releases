# RODENT: getting started

Run your [RODENT website](https://rodent-tau.vercel.app/) experiments from here and save the results back to the website.

**First time only:** click the 🔑 key icon on the left (**Secrets**) and add `RODENT_EMAIL` and `RODENT_PASSWORD` with your website login. Turn on **Notebook access** for both. (If you skip this, the notebook will ask you to type them.)

Then click **Runtime → Run all**.

## 1. Install

```python
%pip install --quiet rodent-sdk==0.4.4
import rodent_sdk as rodent
```

## 2. Sign in

```python
client = rodent.login()

print("Your projects:")
for project in client.projects():
    print(" -", project.name)
```

## 3. Choose your experiment

On the website, open your experiment, click **Use in Python** and copy the code into `EXPERIMENT_CODE`.

Or leave it empty and type the project and experiment names instead.

```python
EXPERIMENT_CODE = ""  #@param {type:"string"}
PROJECT = ""  #@param {type:"string"}
EXPERIMENT = ""  #@param {type:"string"}

if EXPERIMENT_CODE:
    PROJECT, EXPERIMENT = EXPERIMENT_CODE.split("/")

exp = client.project(PROJECT).experiment(EXPERIMENT)
print("Using:", exp.name)
```

### No project yet? (optional)

Type a name, pick an arena and run this cell. Then go back to step 3 and use the new project name. Its experiment is called *your project name* + `baseline`.

```python
NEW_PROJECT = ""  #@param {type:"string"}
ARENA = "complex_habitat"  #@param ["complex_habitat", "mcsf_sprint_one", "open_field", "light_dark_box", "t_maze", "four_room_arena"]

if NEW_PROJECT:
    client.create_project(NEW_PROJECT, template=ARENA)
    print("Created:", NEW_PROJECT)
```

## 4. Look at the setup

```python
print("Seed:     ", exp.seed)
print("Doors:    ", exp.doors)
print("Stimuli:  ", exp.stimuli)
print("Treatment:", exp.treatment)
```

### Change the setup (optional)

Remove the `#` in front of a line and change it. Use the names printed above. Changes are saved to the website.

```python
# exp.seed = 42
# exp.doors["north_gate"].open = False
# exp.stimuli["center_light"].active = True
# exp.treatment.condition = "placebo"

if exp.has_unsaved_changes:
    exp.save()
    print("Saved.")
```

## 5. Run

5000 steps is about 8 minutes in the simulation. The first run takes a minute to download the simulator.

```python
STEPS = 5000  #@param {type:"integer"}

run = exp.run(steps=STEPS)
print("Done:", run)
```

## 6. Results

The rat's path, then the time it spent in each part of the arena.

```python
import matplotlib.pyplot as plt

run.plot()
plt.show()

run.region_table().round(1)
```

Where it spent the most time (brighter = longer):

```python
data = run.to_dataframe()
plt.hist2d(data["x"], data["y"], bins=50, cmap="magma")
plt.gca().invert_yaxis()
plt.colorbar(label="steps")
plt.show()
```

## 7. Save to the website

```python
run.save()
print("Saved. Open the experiment on the website and go to Review to replay it.")
```

## 8. Download the data

```python
run.save_csv("rodent_run.csv")

from google.colab import files
files.download("rodent_run.csv")
```

## Something went wrong?

- **Login failed:** check your two secrets (🔑 key icon) and that Notebook access is on.
- **No project or experiment called ...:** check the spelling, or copy the code from **Use in Python** on the website.
- **Your role does not allow ...:** you can only change some parts of this project. Ask the project lead.
- **Cannot be changed (published):** make a copy first with `exp = exp.save_as_new("New name")`.
